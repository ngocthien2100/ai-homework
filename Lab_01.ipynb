```python
# 1/Nhập 10 số nguyên từ bàn phím, in ra
numbers = []  # Đổi tên biến để tránh trùng với kiểu dữ liệu list
for i in range(10): 
    num = int(input(f"Nhập số nguyên thứ {i + 1}: "))  # Nhập số nguyên
    numbers.append(num)  # Thêm số vào danh sách

print("Danh sách các số đã nhập:", numbers)

# Tính tổng
tong = sum(numbers)
print("Tổng các số:", tong)

# Sắp xếp danh sách
numbers.sort()
print("Danh sách sau khi sắp xếp:", numbers)
```

```python
# 3. Tạo một dictionary chứa thông tin của 5 sinh viên (tên, tuổi, điểm), tính điểm trung bình của tất cả sinh viên.
sinh_vien = {                                           # Tạo dictionary sinh viên
    "SV1": {"tên": "An", "tuổi": 20, "điểm": 8.5},      # Thông tin sinh viên 1
    "SV2": {"tên": "Bình", "tuổi": 21, "điểm": 7.0},    # Thông tin sinh viên 2
    "SV3": {"tên": "Chi", "tuổi": 19, "điểm": 9.0},     # Thông tin sinh viên 3
    "SV4": {"tên": "Dũng", "tuổi": 22, "điểm": 6.5},    # Thông tin sinh viên 4
    "SV5": {"tên": "Em", "tuổi": 20, "điểm": 8.0}       # Thông tin sinh viên 5
}
tong_diem = 0 # Biến lưu tổng điểm
for sv in sinh_vien.values(): # Duyệt qua từng sinh viên
    tong_diem += sv["điểm"] # Cộng điểm của sinh viên vào tổng điểm
diem_trung_binh = tong_diem / len(sinh_vien) # Tính điểm trung bình
print("Điểm trung bình của tất cả sinh viên:", diem_trung_binh) # In ra điểm trung bình
```

```python
# 4. Sử dụng NumPy để tạo mảng 2D (3x3) với các số ngẫu nhiên từ 1 đến 10, tính tổng theo cột và hàng.
import numpy as np # Nhập thư viện NumPy
mang_2d = np.random.randint(1,11,(3,3)) # Tạo mảng 2D 3x3 với số ngẫu nhiên từ 1 đến 10
print("Mang 2D:\n", mang_2d)        # In ra mảng 2D
tong_hang = np.sum(mang_2d, axis=1) # Tính tổng theo hàng
tong_cot = np.sum(mang_2d, axis=0)  # Tính tổng theo cột
print("Tổng theo hàng:", tong_hang) # In ra tổng theo hàng
print("Tổng theo cột:", tong_cot)   # In ra tổng theo cột
```

```python
# 5. Sử dụng Pandas để tạo DataFrame từ danh sách 5 sinh viên, lọc sinh viên có điểm trên 7 và lưu vào file CSV.
import pandas as pd # Nhập thư viện Pandas
data = {            # Tạo dữ liệu cho DataFrame
    "Tên": ["An", "Bình", "Chi", "Dũng", "Em"], # Tên sinh viên
    "Tuổi": [20, 21, 19, 22, 20],               # Tuổi sinh viên  
    "Điểm": [8.5, 7.0, 9.0, 6.5, 8.0]           # Điểm sinh viên
}
df = pd.DataFrame(data)                     # Tạo DataFrame từ dữ liệu
sinh_vien_tren_7 = df[df["Điểm"] > 7]       # Lọc sinh viên có điểm trên 7
print("Sinh viên có điểm trên 7:\n", sinh_vien_tren_7) # In ra sinh viên có điểm trên 7
sinh_vien_tren_7.to_csv("sinh_vien_tren_7.csv", index=False) # Lưu vào file CSV
```

```python
# 6. Sử dụng Matplotlib để vẽ biểu đồ đường cho hàm y = x^3 và biểu đồ cột cho điểm của 5 sinh viên.
import matplotlib.pyplot as plt # Nhập thư viện Matplotlib
# Vẽ biểu đồ đường cho hàm y = x^3
x = np.linspace(-10, 10, 100) # Tạo mảng x
y = x**3                      # Tính y = x^3
plt.figure(figsize=(12, 5))   # Tạo figure với kích thước 12x5
plt.subplot(1, 2, 1)          # Tạo subplot 1
plt.plot(x, y, label='y = x^3', color='blue') # Vẽ biểu đồ đường
plt.title('Biểu đồ đường của y = x^3') # Tiêu đề biểu đồ
plt.xlabel('x')             # Nhãn trục x
plt.ylabel('y')             # Nhãn trục y
plt.legend()                # Hiển thị chú thích
plt.grid()                  # Hiển thị lưới
# Vẽ biểu đồ cột cho điểm của 5 sinh viên
ten_sv = ["An", "Bình", "Chi", "Dũng", "Em"] # Tên sinh viên
diem_sv = [8.5, 7.0, 9.0, 6.5, 8.0]          # Điểm sinh viên
plt.subplot(1, 2, 2)                     # Tạo subplot 2
plt.bar(ten_sv, diem_sv, color='orange') # Vẽ biểu đồ cột
plt.title('Biểu đồ cột điểm của 5 sinh viên') # Tiêu đề biểu đồ
plt.xlabel('Tên sinh viên')    # Nhãn trục x
plt.ylabel('Điểm')             # Nhãn trục y
plt.ylim(0, 10)                # Giới hạn trục y từ 0 đến 10
plt.grid(axis='y')             # Hiển thị lưới theo trục y
plt.tight_layout()             # Tự động điều chỉnh khoảng cách giữa các subplot
plt.show()                     # Hiển thị biểu đồ
```
