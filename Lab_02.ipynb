# 1. Tạo một biến tuple chứa tuổi, giới tính và chiều cao của một người, sau đó thử in ra giới tính của biến tuple đó.
# 2. Tạo một từ điển chứa tuổi, tên, giới tính và số ID của bạn. Lặp từ điển và in ra key -value

````python
class Rectangle:
    def  __init__(self, width, height):
        print(f"Đã tạo một hình chữ nhật với chiều rộng {width} và chiều cao {height}")
        self.width =width
        self.height = height
    
    def setWidth(self, width):
        print(f"Đã thay đổi chiều rộng thành {width}")
        self.width = width
    
    def setHeight(self, height):
        print(f"Đã thay đổi chiều cao thành {height}")
        self.height =height
    
    def getArea(self):
        return self.width * self.height
    
my_rectangle = Rectangle(10,5)

print(f"Diện tích ban đầu: {my_rectangle.getArea()}")

my_rectangle.setHeight(7)

print(f"Diện tích sau khi thay đổi chiều cao: {my_rectangle.getArea()}")
````
